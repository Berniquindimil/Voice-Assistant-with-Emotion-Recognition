## NAIVE BAYES WITH BAG OF WORDS REPRESENTATION

Naïve Bayes is also known as a probabilistic classifier because it is based on Bayes' theorem. This theorem, also known as Bayes' rule, allows us to "invert" conditional probabilities. As a reminder, conditional probabilities represent the probability of an event occurring given that some other event has occurred, which is represented by the following formula:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$
