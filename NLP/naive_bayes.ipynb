## NAIVE BAYES WITH BAG OF WORDS REPRESENTATION

Primero voy a empezar realizando un chatbot simple y a primera vista, de bajo rendimiento, con naïve bayes.

```python
import cv2
import numpy as np
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
```

Aquí podría llamar a la función FER_live_cam() del fichero expression_ssd_detect.py para predecir la emoción facial pero primero voy a realizar una prueba suponiendo que da de output sad.

```python
def get_facial_emotion():
    return "sadness"
```

```python
# Example data: user phrases and corresponding responses (should be trained on a real dataset)
example_texts = [
    "I feel very sad",
    "I am very happy today",
    "Nothing is going right",
    "I am excited about the future",
    "I am so angry",
    "I can't focus",
    "I need urgent help"
]

example_responses = [
    "I'm sorry, it looks like you're feeling sad. Do you want to talk about it?",
    "I'm glad you're feeling happy! Tell me more.",
    "I'm sorry you're going through a tough time. Do you want to share your worries?",
    "It's great to see you excited! What are your plans?",
    "I understand your anger. Sometimes expressing feelings helps.",
    "Lack of focus can be tough. Do you want to explore some tips?",
    "If you feel like you need help, I recommend seeking professional support. I'm here to listen."
]

# Use CountVectorizer to convert text into vectors with Bag of words
vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(example_texts)

# Train a Naïve Bayes classifier (MultinomialNB in this case)
clf = MultinomialNB()
clf.fit(X_train, example_responses)

# Function to generate chatbot response
def generate_response(user_text, facial_emotion):
    # Combine facial emotion and user text for better context
    input_combined = f"{facial_emotion} {user_text}"
    
    # Transform the input text into a vector
    X_input = vectorizer.transform([input_combined])
    
    # Predict the response using the Naïve Bayes classifier
    predicted_response = clf.predict(X_input)[0]
    return predicted_response

# --- Example chatbot flow integration ---

# Suppose the user inputs a text message
user_message = input("How could I help you?")

# Get detected facial emotion (this should come from FER_live_cam in your real implementation)
facial_emotion = get_facial_emotion()

# Generate chatbot response considering detected emotion and user message
chatbot_response = generate_response(user_message, facial_emotion)
print("Chatbot Response:", chatbot_response)
```
